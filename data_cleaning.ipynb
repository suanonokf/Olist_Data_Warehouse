```python
import pandas as pd
```

1-Remove Undesirable columns

```python
orders_df = pd.read_csv('./raw_dataset/olist_orders_dataset.csv')
orders_df=orders_df.drop(columns=['order_approved_at','order_delivered_carrier_date','order_estimated_delivery_date'])
orders_df.head()
```

```python
customers_df = pd.read_csv('./raw_dataset/olist_customers_dataset.csv')
customers_df = customers_df.drop(columns=['customer_city','customer_state'])
customers_df.head()
```

```python
products_df = pd.read_csv('./raw_dataset/olist_products_dataset.csv')
products_df=products_df.drop(columns=['product_name_lenght','product_description_lenght','product_photos_qty'])
products_df.head()
```

```python
reviews_df = pd.read_csv('./raw_dataset/olist_order_reviews_dataset.csv')
reviews_df = reviews_df.drop(columns=['review_comment_title','review_comment_message','review_creation_date','review_answer_timestamp'])
reviews_df.head()
```

```python
sellers_df = pd.read_csv('./raw_dataset/olist_sellers_dataset.csv')
sellers_df=sellers_df.drop(columns=['seller_city','seller_state'])
sellers_df.head()
```

```python
geolocation_df = pd.read_csv('./raw_dataset/olist_geolocation_dataset.csv')
geolocation_df = geolocation_df.drop(columns=['geolocation_lat','geolocation_lng'])
geolocation_df.head()
```
